```python
import pandas as pd
import json

df = pd.read_json("../data/evaluation/test_queries.jsonl", lines=True)
from pathlib import Path


PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

EMBEDDINGS_DIR = PROJECT_ROOT / "data/embeddings"
CHUNKS_DIR = PROJECT_ROOT / "data/chunks"
```

```python
for query in df:
    print(query)
```

```python
chunks = {}
for path in CHUNKS_DIR.glob("*/*.chunks.jsonl"):
    for line in path.open():
        chunk = json.loads(line)
        chunks[chunk["chunk_id"]] = chunk["text"]

df_with_gold_answers = df.copy()
df_with_gold_answers["gold_chunk_answer"] = df_with_gold_answers["expected_chunk_ids"].apply(
    lambda ids: "\n\n".join(f"{chunk_id}: {chunks[chunk_id]}" for chunk_id in ids)

)
df_with_gold_answers.to_json('test_queries_with_chunks.jsonl', orient='records', indent=4)
```
